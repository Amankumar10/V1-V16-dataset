```python
import json

input_file = "./dataset/V1A/V1A_large_8.1k.jsonl"
output_file = "./leaned_dataset.jsonl"

with open(input_file, "r", encoding="utf-8") as fin, \
     open(output_file, "w", encoding="utf-8") as fout:

    for line in fin:
        if not line.strip():
            continue

        item = json.loads(line)

        item.get("input", {}).pop("domain", None)
        item.get("input", {}).pop("relationship", None)

        fout.write(json.dumps(item, ensure_ascii=False) + "\n")

print(f"Cleaned dataset saved to {output_file}")
```
