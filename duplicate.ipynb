```python
import json
from collections import defaultdict

FILE = "./dataset/V3A/V3Anew.jsonl"

seen = {}
duplicates = defaultdict(list)

total = 0

with open(FILE, "r", encoding="utf-8") as f:
    for line_num, line in enumerate(f, 1):
        line = line.strip()

        if not line:
            continue

        total += 1

        data = json.loads(line)

        # Normalize JSON so key order doesn't matter
        key = json.dumps(
            data,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":")
        )

        if key in seen:
            duplicates[key].append(line_num)
        else:
            seen[key] = line_num


duplicate_groups = len(duplicates)
duplicate_records = sum(len(lines) for lines in duplicates.values())

print("=" * 60)
print("DUPLICATE CHECK")
print("=" * 60)

print(f"Total samples          : {total}")
print(f"Unique samples         : {len(seen)}")
print(f"Duplicate groups       : {duplicate_groups}")
print(f"Duplicate records      : {duplicate_records}")

if total:
    print(f"Duplicate percentage   : {duplicate_records / total * 100:.2f}%")

print("\n===== Duplicate Locations =====")

if not duplicates:
    print("No exact duplicate samples found.")
else:
    for i, (key, lines) in enumerate(duplicates.items(), 1):
        original_line = seen[key]

        print(f"\nDuplicate Group {i}")
        print(f"Original line : {original_line}")
        print(f"Duplicates    : {lines}")
        print(f"Total copies  : {len(lines) + 1}")
```
