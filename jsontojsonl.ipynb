```python

import json
from pathlib import Path

# Configuration
INPUT_FILE = Path("new.jsonl")
OUTPUT_FILE = Path("output.jsonl")


def convert_json_to_jsonl(input_file, output_file):
    total = 0
    success = 0
    failed = 0

    print("Starting JSON to JSONL conversion...")

    try:
        with open(input_file, "r", encoding="utf-8") as f:
            data = json.load(f)

        # Support both JSON arrays and single JSON objects
        if isinstance(data, dict):
            data = [data]

        if not isinstance(data, list):
            raise ValueError("JSON must contain an array or object.")

        with open(output_file, "w", encoding="utf-8") as out:
            for index, item in enumerate(data, start=1):
                total += 1

                try:
                    json_line = json.dumps(
                        item,
                        ensure_ascii=False,
                        separators=(",", ":")
                    )

                    # Validate generated JSON
                    json.loads(json_line)

                    out.write(json_line + "\n")
                    success += 1

                except (TypeError, ValueError) as e:
                    failed += 1
                    print(f"Record {index} skipped: {e}")

                if total % 1000 == 0:
                    print(f"Processed: {total}")

        print("\nConversion completed!")
        print(f"Total records: {total}")
        print(f"Successful: {success}")
        print(f"Failed: {failed}")
        print(f"Output: {output_file.resolve()}")

    except json.JSONDecodeError as e:
        print(f"Invalid JSON file: {e}")

    except Exception as e:
        print(f"Error: {e}")


if __name__ == "__main__":
    convert_json_to_jsonl(INPUT_FILE, OUTPUT_FILE)
```
